```python
test_prompt = "You need to write a python function with 2 inputs: list of numbers and expected sum. Function output format: list of tuples. Function needs to find all groups by 3 numbers from the list which sum is equal to expected sum variable. Each set from 3 numbers should be presented only once without permutations. Estimate the computational complexity of the solution; if there are several solutions, implement the one with the minimum computational complexity."
```

Anthropic Claude Opus 4.6 solution

```python
def find_triplets(numbers, expected_sum):
    """
    Find all unique triplets in the list that sum up to the expected sum.
    
    Args:
        numbers: List of numbers
        expected_sum: Target sum for triplets
        
    Returns:
        List of tuples, each containing 3 numbers that sum to expected_sum
    
    Computational Complexity: O(n²) time, O(n) space (excluding output)
    """
    numbers.sort()  # O(n log n)
    result = []
    n = len(numbers)
    
    for i in range(n - 2):
        # Skip duplicate values for the first element
        if i > 0 and numbers[i] == numbers[i - 1]:
            continue
        
        # Two-pointer technique for the remaining two elements
        left = i + 1
        right = n - 1
        
        while left < right:
            current_sum = numbers[i] + numbers[left] + numbers[right]
            
            if current_sum == expected_sum:
                result.append((numbers[i], numbers[left], numbers[right]))
                
                # Skip duplicates for the second element
                while left < right and numbers[left] == numbers[left + 1]:
                    left += 1
                # Skip duplicates for the third element
                while left < right and numbers[right] == numbers[right - 1]:
                    right -= 1
                    
                left += 1
                right -= 1
            elif current_sum < expected_sum:
                left += 1
            else:
                right -= 1
    
    return result
```

Gemini 3.1 solution

```python
def find_triplets_with_sum(nums, expected_sum):
    # Sort the array first to easily manage duplicates and use two pointers
    nums.sort()
    n = len(nums)
    triplets = []
    
    for i in range(n - 2):
        # Skip duplicate elements for the first number to avoid duplicate triplets in the output
        if i > 0 and nums[i] == nums[i - 1]:
            continue
            
        left = i + 1
        right = n - 1
        
        while left < right:
            current_sum = nums[i] + nums[left] + nums[right]
            
            if current_sum == expected_sum:
                triplets.append((nums[i], nums[left], nums[right]))
                
                # Skip duplicate elements for the second number
                while left < right and nums[left] == nums[left + 1]:
                    left += 1
                # Skip duplicate elements for the third number
                while left < right and nums[right] == nums[right - 1]:
                    right -= 1
                    
                # Move both pointers inward after finding a valid triplet
                left += 1
                right -= 1
                
            elif current_sum < expected_sum:
                # If the sum is too small, we need a larger number, so move the left pointer to the right
                left += 1
            else:
                # If the sum is too large, we need a smaller number, so move the right pointer to the left
                right -= 1
                
    return triplets

# Example usage:
# numbers = [-1, 0, 1, 2, -1, -4, 3]
# target = 2
# print(find_triplets_with_sum(numbers, target))
# Output: [(-1, 0, 3), (-1, 1, 2)]
```

GPT 5.5 Solution

```python
from typing import List, Tuple

def three_number_groups(numbers: List[int], expected_sum: int) -> List[Tuple[int, int, int]]:
    """
    Returns all unique triplets of numbers whose sum equals expected_sum.
    Each triplet appears only once, without permutations.
    """

    nums = sorted(numbers)
    result = []

    n = len(nums)

    for i in range(n - 2):
        # Skip duplicate first elements
        if i > 0 and nums[i] == nums[i - 1]:
            continue

        left = i + 1
        right = n - 1

        while left < right:
            current_sum = nums[i] + nums[left] + nums[right]

            if current_sum == expected_sum:
                result.append((nums[i], nums[left], nums[right]))

                left += 1
                right -= 1

                # Skip duplicates for second element
                while left < right and nums[left] == nums[left - 1]:
                    left += 1

                # Skip duplicates for third element
                while left < right and nums[right] == nums[right + 1]:
                    right -= 1

            elif current_sum < expected_sum:
                left += 1
            else:
                right -= 1

    return result
```
