```python
from pptx import Presentation
from pptx.util import Inches, Pt
from pptx.enum.text import PP_ALIGN
from pptx.dml.color import RGBColor

# Create presentation
prs = Presentation()
prs.slide_width = Inches(10)
prs.slide_height = Inches(7.5)

# Define color scheme
COLOR_PRIMARY = RGBColor(0, 51, 102)  # Dark blue
COLOR_ACCENT = RGBColor(220, 20, 60)  # Crimson
COLOR_TEXT = RGBColor(51, 51, 51)  # Dark gray
COLOR_LIGHT = RGBColor(240, 240, 240)  # Light gray

# Slide data
slides_data = [
    {
        "title": "The Omnichannel Inventory Crisis",
        "body": "Meridian operates 600 stores + online across 8 countries with fragmented inventory systems. No real-time visibility between channels. Competitors (Zalando, ASOS) have unified inventory; Meridian is losing customers.",
        "speaker_note": "We're losing market share because our inventory systems can't talk to each other. Customers see phantom stock online, drive to stores, find nothing. Zalando solved this; we haven't."
    },
    {
        "title": "12% Phantom Stock = €8M/Year Lost Margin",
        "body": "Click-&-collect orders cancelled at pickup: 12% rate. Customers explicitly churn to Zalando/ASOS. Store staff manually retype loyalty IDs. Regional planners use spreadsheets. No cross-channel visibility.",
        "speaker_note": "This isn't a nice-to-have. Customers are leaving. We have primary evidence: app reviews, support forums, Reddit posts all say the same thing: 'I switched to Zalando.'"
    },
    {
        "title": "Demand-Driven Inventory Allocation",
        "body": "AI-powered demand forecasting + store-level allocation optimization. Predict demand 7–14 days ahead. Recommend where inventory should live (store vs. online). Planners approve/reject. Reduce phantom stock to ≤5%.",
        "speaker_note": "This is not a new idea. Zalando, ASOS, Inditex all do this. We're 2 years behind. The technology is proven; the question is execution."
    },
    {
        "title": "How It Works",
        "body": "Daily: forecast demand per SKU per store. Calculate optimal allocation. Generate recommendations ranked by impact. Planners approve via dashboard. System integrates with SAP. Weekly: measure phantom stock reduction, retrain model.",
        "speaker_note": "The system is simple: forecast, optimize, recommend, approve, execute, measure. No autonomous decisions. Planners stay in control. We're automating the spreadsheet, not replacing the planner."
    },
    {
        "title": "Base Case: €2.3M Year 1 Value",
        "body": "Phantom stock margin saved: €900K. Customer churn reduction: €675K. Planner time freed: €750K. Total Year 1 value: €2.325M. Payback: 4.5 months. ROI: 265% Year 1.",
        "speaker_note": "These numbers are conservative. We're using industry benchmarks from Gartner, McKinsey, Deloitte. Even in the pessimistic case, payback is 24 months. This is a no-brainer."
    },
    {
        "title": "Three Scenarios: Payback Period",
        "body": "Pessimistic (slow adoption): 24–30 months. Base case (moderate adoption): 4.5 months. Optimistic (fast adoption): 1.1 months. Implementation cost: €879K (base). Year 1 run cost: €180K.",
        "speaker_note": "Even pessimistic assumes we execute well. If we pilot in 1–2 countries first, measure results, then scale, we de-risk the pessimistic scenario significantly."
    },
    {
        "title": "Top-2 Value Drivers",
        "body": "Driver 1: Phantom stock reduction rate (±29% impact on value). Driver 2: Planner adoption rate (±21% impact). If adoption is slow, value drops. Mitigation: pilot first, measure weekly.",
        "speaker_note": "These two drivers are where we focus. If planners don't trust the system, it fails. If phantom stock doesn't actually drop, it fails. We need to de-risk both."
    },
    {
        "title": "Binding Gate: Planner Adoption",
        "body": "Risk: Planners ignore recommendations (organizational resistance, complexity, distrust). Gate: Pilot in 1–2 countries. Success threshold: ≥70% adoption rate within 3 months. If <70%, pause rollout, investigate.",
        "speaker_note": "This is the make-or-break risk. We can build the best model in the world, but if planners don't use it, we fail. Change management is non-negotiable."
    },
    {
        "title": "Investment & Timeline",
        "body": "Budget: €879K implementation + €180K Year 1 run cost = €1.059M total Year 1. Timeline: 6 months to GA (Phase 1: pilot 1–2 countries; Phase 2: rollout).",
        "speaker_note": "We need €879K upfront, €180K for operations. That's 0.01% of annual revenue. Payback in 4.5 months. We need executive sponsorship to push through organizational resistance."
    },
    {
        "title": "30-Day Roadmap",
        "body": "Week 1: Validate CRM data (click-&-collect volume, churn rate). Week 2: Secure pilot countries (Italy + Germany). Week 3: Kick-off pilot. Week 4: First forecast model.",
        "speaker_note": "We move fast. First 30 days are about de-risking assumptions and building momentum. By end of Month 1, we'll have a working forecast model in pilot countries."
    }
]

def add_slide_with_title_and_body(prs, title, body, speaker_note):
    """Add a slide with title, body text, and speaker notes"""
    
    # Add blank slide
    blank_slide_layout = prs.slide_layouts[6]  # Blank layout
    slide = prs.slides.add_slide(blank_slide_layout)
    
    # Add background color
    background = slide.background
    fill = background.fill
    fill.solid()
    fill.fore_color.rgb = RGBColor(255, 255, 255)  # White background
    
    # Add title box
    title_left = Inches(0.5)
    title_top = Inches(0.5)
    title_width = Inches(9)
    title_height = Inches(1)
    
    title_box = slide.shapes.add_textbox(title_left, title_top, title_width, title_height)
    title_frame = title_box.text_frame
    title_frame.word_wrap = True
    
    title_paragraph = title_frame.paragraphs[0]
    title_paragraph.text = title
    title_paragraph.font.size = Pt(44)
    title_paragraph.font.bold = True
    title_paragraph.font.color.rgb = COLOR_PRIMARY
    title_paragraph.alignment = PP_ALIGN.LEFT
    
    # Add horizontal line under title
    line_shape = slide.shapes.add_shape(
        1,  # Line shape
        Inches(0.5),
        Inches(1.6),
        Inches(9),
        Inches(0)
    )
    line_shape.line.color.rgb = COLOR_ACCENT
    line_shape.line.width = Pt(2)
    
    # Add body text box
    body_left = Inches(0.5)
    body_top = Inches(2)
    body_width = Inches(9)
    body_height = Inches(4.5)
    
    body_box = slide.shapes.add_textbox(body_left, body_top, body_width, body_height)
    body_frame = body_box.text_frame
    body_frame.word_wrap = True
    
    body_paragraph = body_frame.paragraphs[0]
    body_paragraph.text = body
    body_paragraph.font.size = Pt(24)
    body_paragraph.font.color.rgb = COLOR_TEXT
    body_paragraph.alignment = PP_ALIGN.LEFT
    body_paragraph.line_spacing = 1.5
    
    # Add speaker notes
    notes_slide = slide.notes_slide
    text_frame = notes_slide.notes_text_frame
    text_frame.text = speaker_note
    
    # Add slide number at bottom
    slide_number_left = Inches(9)
    slide_number_top = Inches(7)
    slide_number_width = Inches(0.5)
    slide_number_height = Inches(0.3)
    
    slide_number_box = slide.shapes.add_textbox(slide_number_left, slide_number_top, slide_number_width, slide_number_height)
    slide_number_frame = slide_number_box.text_frame
    slide_number_paragraph = slide_number_frame.paragraphs[0]
    slide_number_paragraph.text = str(len(prs.slides))
    slide_number_paragraph.font.size = Pt(12)
    slide_number_paragraph.font.color.rgb = COLOR_LIGHT
    slide_number_paragraph.alignment = PP_ALIGN.RIGHT

# Add title slide
title_slide_layout = prs.slide_layouts[6]  # Blank layout
title_slide = prs.slides.add_slide(title_slide_layout)

# Title slide background
background = title_slide.background
fill = background.fill
fill.solid()
fill.fore_color.rgb = COLOR_PRIMARY

# Title slide content
title_left = Inches(0.5)
title_top = Inches(2.5)
title_width = Inches(9)
title_height = Inches(2)

title_box = title_slide.shapes.add_textbox(title_left, title_top, title_width, title_height)
title_frame = title_box.text_frame
title_frame.word_wrap = True

title_paragraph = title_frame.paragraphs[0]
title_paragraph.text = "Demand-Driven Inventory Allocation"
title_paragraph.font.size = Pt(54)
title_paragraph.font.bold = True
title_paragraph.font.color.rgb = RGBColor(255, 255, 255)
title_paragraph.alignment = PP_ALIGN.CENTER

# Subtitle
subtitle_left = Inches(0.5)
subtitle_top = Inches(4.5)
subtitle_width = Inches(9)
subtitle_height = Inches(1)

subtitle_box = title_slide.shapes.add_textbox(subtitle_left, subtitle_top, subtitle_width, subtitle_height)
subtitle_frame = subtitle_box.text_frame
subtitle_frame.word_wrap = True

subtitle_paragraph = subtitle_frame.paragraphs[0]
subtitle_paragraph.text = "Meridian Retail Group | EU Fashion Omnichannel"
subtitle_paragraph.font.size = Pt(28)
subtitle_paragraph.font.color.rgb = RGBColor(255, 255, 255)
subtitle_paragraph.alignment = PP_ALIGN.CENTER

# Add content slides
for i, slide_data in enumerate(slides_data, 1):
    add_slide_with_title_and_body(
        prs,
        f"Slide {i}: {slide_data['title']}",
        slide_data['body'],
        slide_data['speaker_note']
    )

# Save presentation
output_file = "06-deck.pptx"
prs.save(output_file)
print(f"✅ Deck created: {output_file}")
print(f"📊 Total slides: {len(prs.slides)}")
print(f"📝 Ready to export as PDF")

# Print summary
print("\n" + "="*60)
print("DECK SUMMARY")
print("="*60)
for i, slide_data in enumerate(slides_data, 1):
    word_count = len(slide_data['body'].split())
    print(f"Slide {i}: {word_count} words ✅" if word_count <= 30 else f"Slide {i}: {word_count} words ⚠️")
print("="*60)
```
